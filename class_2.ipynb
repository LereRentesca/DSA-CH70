#### Mini Challenge 1: Largest Num
Create a function that will help us to determine the largest number from a list of numbers

Criteria:
1. The function should receive a list and return the largest number
2. You can test it with a list of any values from your selection

```python
numbers = [230,2,3,10,23,89,11]

def find_largest_number(numbers):
    largest = numbers[0]
    for number in numbers[1:]:
        if number > largest:
            largest = number

    return largest

print(f"The largest number is: {find_largest_number(numbers)}")
```

### Mini Challenge 2: Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagrams of each other. False otherwise.

    Example:
    1. heart and earth
    2. scar and cars
    3. star and rats

*Criteria:*  
Two strings are anagrams of each other if and only if all of the criteria below are met:

    1. They must use all the same characters
    2. They must be of exactly the same length (in chars)
    3. AI and sorted function is not allowed


```python
def are_anagrams(str1, str2): # O(1)
    counter = 0 # O(1)
    if len(str1) == len(str2): # O(1)
        for letter1 in str1: # O(n)
            for letter2 in str2: # O(n) -> O(n^2)
                if letter1 == letter2: # O(1)
                    counter += 1 # O(1)
        if len(str1) != counter: # O(1)
            return False
        else:
            return True
    else:    
        return False
```

```python
class Calculator:

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        result = self.num1 + self.num2
        return result
    
    def subtract(self):
        return self.num1 - self.num2
    
    def multiply(self):
        return self.num1 * self.num2
    
    def divide(self):
        if self.num2 == 0:
            raise ArithmeticError("Element can't be divide by 0")
        return self.num1 / self.num2
    
    def __str__(self):
        return f"Calculator (num1={self.num1}, num2={self.num2})"
```

```python
c1 = Calculator(20, 5)
print(c1.add())
print(c1.subtract())
print(c1.multiply())
print(c1.divide())
print(c1)
```
