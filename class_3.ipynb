## Stacks
A stack is an ordered collection of elements where items are added and removed from one end, commonly referred to as the "top". The ordering principle is referred to as LIFO (Last In, First Out)

```python
# Simplified Version (relying on built-ins functions)

class Stack:
    def __init__(self):
        self.items = []

    def push(self, elements):
        self.items.append(elements)

    def push(self, *args):
        for item in args:
            self.items.append(item)

    def pop(self):
        return self.items.pop()
    
    def peek(self):
        return self.items[-1]
    
    def is_empty(self):
        return self.items == []
    
    def size(self):
        return len(self.items)
    
    def __str__(self):
        return f"Stack - {self.items}"
```

```python
s1 = Stack()
print(s1)
print(s1.is_empty())
s1.push("A")
s1.push("B")
s1.push("C")
print(s1.size())
print(s1.peek())
print(s1.is_empty())
top = s1.pop()
print(s1.size())
print(top)
print(s1)
#s1.push(["C","D"])
s1.push("C", "D","E", "F")
print(s1)
```

```python
# CustomStack (from scratch)

class CustomStack:
    def __init__(self): # Empty constructor for CustomStack
        self.top = None # This class will contain a single attribute called top

    class Node: # Inner class living inside of 'CustomStack'
        def __init__(self, data): # Constructor for Node, will require a single attribute called data to create an obj
            self.data = data 
            self.below = None # It will also contain a second attribute called below (This one is not mandatory in
                                # the creation of the object)

        def __str__(self): # To String helper method to convert obj into something readable
            return f"Node - {self.data}"
        
    def is_empty(self): # If the top attribute is None then our Stack is empty
        return self.top is None
    
    def push(self, element): # Push method will help us to add elements to the stack
        if self.is_empty(): # First Scenario: If it's empty I need to set the top reference to the new node
            self.top = self.Node(element)
        else: # Second Scenario: If it's not empty, then I need to do the following
            new_node = self.Node(element) # 1. Create a new node for the data I want to store
            new_node.below = self.top # 2. Link the NEW node to the previous one (CURRENT top)
            self.top = new_node # 3. Update the top reference to the new node

    def pop(self): # Pop method will remove the top node and return the value
        if self.is_empty(): # First Scenario: If it's empty I will raise a ReferenceError because there is no data to retrieve
            raise ReferenceError("Stack is empty")
        else: # Second Scenario: If it's not empty then:
            value_to_remove = self.top.data # 1. Create a variable with the top value
            self.top = self.top.below # 2. Update the top reference to the previous node (the BELOW one from the current top)
            return value_to_remove # 3. Return the variable with the value of the removed node

    def size(self):
        if self.is_empty(): 
            raise ReferenceError("Stack is empty")
        else:
            count = 0 # Create a variable to keep the counts of the nodes
            current_node = self.top # Create a variable of the current node so we can start traveling through the nodes
            while not current_node is None: # We will iterate if the current node is not None, once the node is None we will stop
                count += 1 # Each iteration we will increase the counts
                current_node = current_node.below # And we will move to the node below
            return count # Once we reach the node with the 'None' below, let's return the count variable

    def __str__(self):
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            append_node = "" # Create a variable to concatenate each value from the node
            current_node = self.top # Create a copy of the stack to travel across the nodes
            while not current_node is None: # We will iterate if the current node is not None, once the node is None we will stop
                append_node += f"{current_node.data}, " # Each iteration we will take the data from the node and append it to the variable
                current_node = current_node.below # And we will move to the node below
            return f"CustomStack - [{append_node}]" # At the end, return a string with the title of the class with the appended variable

    def peek(self):
        # Should return the value from the top node, not taking it away or removing it from the stack, just displaying the value.
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            return self.top.data
    
```

```python
customStack1 = CustomStack()
print(customStack1.is_empty())
customStack1.push("A")
customStack1.push("B")
customStack1.push("C")
print(customStack1.size())
print(customStack1)
```

Use the Stack class to invert a string  
Example:  

    given "Rafael" your function should return "leafaR"
    given "stars" your function should return "srats"
    given "cool" your function should return "looc"  

**Your function must receive a string as a parameter and return a string(not a list)**  
**Your solution must include the built-in Stack class created in class**  
**You cannot google anything for this challenge**  

```python
def invert_string(word):
    inverted_word = ""

    stack = Stack()
    for letter in word:
        stack.push(letter)

    while not stack.is_empty():
        inverted_word += stack.pop()

    return inverted_word

print(invert_string("Hello"))
print(invert_string("Rafael"))
print(invert_string("Cool"))
```
