# OOP (Object Oriented Programming)
Python is an object-oriented programming, allowing us to structure code using classes and objects 
for a better organization and reusability.

## Advantages
- Provides clear structure to programs
- Makes code easier to mantain, reuse and debug
- Helps keep your code DRY (Do not Repeat Yourself principle)
- Allows to rebuild reusable applications with less code

## Classes and Objects
A class defines what an object should look like and it's behavior  
An object is created based on that class  
For example:  

    Class -> Fruit
    Objects -> Apple, Banana, Mango
    Class -> Car
    Objects -> Audi, Tesla, Ford

#### The four pillars of OOP
1. Encapsulation
2. Abstraction
3. Inheritance
4. Polymorphism

Example of a class:  
    class PostListView(ListView):

```python
class Person:
    # Constructor
    def __init__(self, name, last_name, age):
        self.name = name
        self.last_name = last_name
        self.age = age
        self.is_active = False

    def greeting(self):
        return f"Hello from {self.name}"
    
    def __str__(self):
        return f"Person (name = {self.name}, last_name = {self.last_name}, age = {self.age}, is_active = {self.is_active})"
    
p1 = Person("Luis", "Renteria", 25)
p1.is_active = True
print(p1)
print(p1.greeting())

p2 = Person("Lina", "Holland", 29)
print(p2)
print(p2.greeting())
```

# Assignment 1: Find Even Numbers

Find the total of numbers from a collection that divide by 2 the modulus is 0

Examples:
1. Given [3,4,10,15,999,2] Then the output should be: total = 3  
2. Given [2,4,6,10,3.7] Then the output should be: total = 4

Criteria:
1. Your function should receive a list and return an integer

Hint:

    - '%' is the math operand to get the modulus from a division
    - 2 % 2 = 0

The collection is:  
numbers = [3,5,27,20,6,3.5,60]

```python
def count_even_numbers(collection):
    total = 0
    for number in collection:
        if number % 2 == 0:
            total += 1
    
    return total

numbers = [3,5,27,20,6,3.5,60]
print(f"The total is: {count_even_numbers(numbers)}")
```
